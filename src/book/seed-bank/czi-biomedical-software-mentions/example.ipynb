```python
# used doi_10_5061_dryad_zgmsbccjk__v20241112.zip
# obtained from https://datadryad.org/dataset/doi:10.5061/dryad.zgmsbccjk
# stored under data dir within same directory as this file.

import duckdb


with duckdb.connect() as conn:
    df_disambiguation = conn.execute(
        """
        WITH disambig_software AS (
            SELECT DISTINCT 
                software
            FROM "doi_10_5061_dryad_zgmsbccjk__v20241112/mentions/disambiguation_files/final_output/*.csv.gz"
        )
        SELECT
            ID,
            software_mention,
            mapped_to,
            source,
            platform,
            package_url,
            description,
            homepage_url,
            license,
            github_repo
        FROM "doi_10_5061_dryad_zgmsbccjk__v20241112/mentions/linked/metadata.tsv.gz"
        WHERE 
            software_mention IN (SELECT software FROM disambig_software)
            AND github_repo IS NOT NULL
            AND github_repo <> ''
            AND exact_match = True
    """
    ).df()
df_disambiguation
```

```python
df_disambiguation.dropna(subset=["github_repo"]).to_parquet("czi_disambiguated_biomedical_software_github_refs.parquet", index=False)
```
